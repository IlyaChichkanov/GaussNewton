# Постановка задачи оптимизации Single shooting

Пусть динамика системы задаётся дифференциальным уравнением

$$
\begin{equation}
    \dot{x} = f(t, x, \theta),\quad x(t_0) = c_0,
\end{equation}
$$

где 
- $t \in [t_0, t_f]$ — время,
- $x \in \mathbb{R}^{n_x}$ — вектор состояния системы,
- $\theta \in \mathbb{R}^{n_\theta}$ — вектор постоянных параметров,
- $c_0 \in \mathbb{R}^{n_x}$ — начальное состояние,
- $f: \mathbb{R}^{n_x} \times \mathbb{R}^{n_\theta} \to \mathbb{R}^{n_x}$ — непрерывно дифференцируема по $x$ и $\theta$.

Положим *неизвестными* и *независимыми* вектор параметров $\theta$ и начальное состояние $c_0$.

На временном интервале $[t_0, t_f]$ имеется $N$ измерений:

$$
y_i = h(x(t_i), \theta) + \varepsilon_i, \quad i = 1,\dots,N,
$$

где 
- $h: \mathbb{R}^{n_x} \times \mathbb{R}^{n_\theta} \to \mathbb{R}^{n_y}$ — функция измерений (непрерывно дифференцируема),
- $y_i \in \mathbb{R}^{n_y}$ — измерение,
- $\varepsilon_i \sim \mathcal{N}(0, \Sigma_i)$ — независимые гауссовские ошибки с ковариацией $\Sigma_i$.

**Цель:** найти $\hat\theta$ и $\hat c_0$, минимизирующие взвешенную сумму квадратов невязок.

Обозначим через $x(t \mid \theta, c_0)$ решение (1) в момент $t$ при заданных $\theta, c_0$.

Целевая функция:
$$
S(\theta, c_0) = \frac12 \sum_{i=1}^{N} \bigl( h(x(t_i \mid \theta, c_0), \theta) - y_i \bigr)^T W_i \bigl( h(x(t_i \mid \theta, c_0), \theta) - y_i \bigr),
$$
где $W_i = \Sigma_i^{-1}$ — весовые матрицы (обратные ковариациям шума).

Задача:
$$
\min_{\theta, c_0} S(\theta, c_0) \quad \text{при условии} \quad \dot{x} = f(t,x,\theta),\; x(t_0)=c_0.
$$


## Вариация целевой функции

Рассмотрим вариацию $S$ относительно $\theta$ и $c_0$.

Линеаризация решения:
$$
x(t \mid \theta+\delta\theta, c_0+\delta c_0) \approx x(t \mid \theta, c_0) + J_\theta(t) \delta\theta + J_{c_0}(t) \delta c_0,
$$
где
$$
J_\theta(t) = \frac{\partial x(t \mid \theta, c_0)}{\partial \theta} \in \mathbb{R}^{n_x \times n_\theta},\qquad
J_{c_0}(t) = \frac{\partial x(t \mid \theta, c_0)}{\partial c_0} \in \mathbb{R}^{n_x \times n_x}.
$$


## Вывод вариационных уравнений

Дифференцируем уравнение движения по $\theta$:

$$
\frac{\partial}{\partial \theta} \dot{x}(t) = \frac{\partial}{\partial \theta} f(t, x(t), \theta).
$$

Левая часть: меняем порядок дифференцирования

$$
\frac{\partial}{\partial \theta} \dot{x}(t) = \frac{d}{dt} \left( \frac{\partial x(t)}{\partial \theta} \right) = \dot{J}_\theta(t).
$$

Правая часть: по правилу дифференцирования сложной функции

$$
\frac{\partial}{\partial \theta} f(t, x(t), \theta) = \frac{\partial f}{\partial x} \cdot \frac{\partial x(t)}{\partial \theta} + \frac{\partial f}{\partial \theta}
= f_x(t,x,\theta)\, J_\theta(t) + f_\theta(t,x,\theta).
$$

Получаем дифференциальное уравнение:

$$
\dot{J}_\theta(t) = f_x(t,x,\theta)\, J_\theta(t) + f_\theta(t,x,\theta).
$$

Начальное условие: $x(t_0)=c_0$ не зависит от $\theta$, поэтому

$$
J_\theta(t_0) = \frac{\partial c_0}{\partial \theta} = \mathbf{0}_{n \times p}.
$$

### 2. Чувствительность по начальным условиям $c_0$

Определим

$$
J_{c_0}(t) = \frac{\partial x(t)}{\partial c_0} \in \mathbb{R}^{n \times n}.
$$

Дифференцируем уравнение по $c_0$:

$$
\frac{\partial}{\partial c_0} \dot{x}(t) = \frac{\partial}{\partial c_0} f(t, x(t), \theta).
$$

Левая часть:

$$
\frac{\partial}{\partial c_0} \dot{x}(t) = \frac{d}{dt} \left( \frac{\partial x(t)}{\partial c_0} \right) = \dot{J}_{c_0}(t).
$$

Правая часть: $f$ зависит от $c_0$ только через $x(t)$

$$
\frac{\partial}{\partial c_0} f(t, x(t), \theta) = \frac{\partial f}{\partial x} \cdot \frac{\partial x(t)}{\partial c_0} = f_x(t,x,\theta)\, J_{c_0}(t).
$$

Отсюда

$$
\dot{J}_{c_0}(t) = f_x(t,x,\theta)\, J_{c_0}(t).
$$

Начальное условие: $x(t_0)=c_0$, следовательно

$$
J_{c_0}(t_0) = \frac{\partial c_0}{\partial c_0} = I_n.
$$

### 3. Итоговые вариационные уравнения

Собирая полученные результаты:

$$
\begin{aligned}
\frac{d}{dt} J_\theta &= f_x(t,x,\theta)\, J_\theta + f_\theta(t,x,\theta), \quad & J_\theta(t_0) = 0,\\[4pt]
\frac{d}{dt} J_{c_0} &= f_x(t,x,\theta)\, J_{c_0}, \quad & J_{c_0}(t_0) = I_{n_x}.
\end{aligned}
$$

Эти уравнения позволяют численно вычислять производные решения по параметрам и начальным условиям, что необходимо при решении задач оптимизации (идентификация параметров, MHE) методом Гаусса–Ньютона.


Линеаризация функции измерений:
$$
h\bigl(x(t)+\delta x, \theta+\delta\theta\bigr) \approx h(x(t),\theta) + h_x \delta x + h_\theta \delta\theta,
$$
где $h_x = \frac{\partial h}{\partial x}$, $h_\theta = \frac{\partial h}{\partial \theta}$.

Подставляя $\delta x = J_\theta\delta\theta + J_{c_0}\delta c_0$, получаем:
$$
h(\cdot) \approx h(x(t),\theta) + \bigl( h_x J_\theta + h_\theta \bigr)\delta\theta + h_x J_{c_0}\delta c_0.
$$

Тогда невязка:
$$
r_i(\theta,c_0) = y_i - h(x(t_i),\theta).
$$

Вариация целевой функции (с весами):
$$
S(\theta+\delta\theta,c_0+\delta c_0) \approx \frac12 \sum_{i=1}^N \bigl\| \bigl( h_x J_\theta + h_\theta \bigr)\delta\theta + h_x J_{c_0}\delta c_0 - r_i \bigr\|_{W_i}^2.
$$

## Метод Гаусса-Ньютона

На $k$-й итерации имеем $\theta^k$, $c_0^k$. Решаем линейную задачу наименьших квадратов:

$$
\min_{\delta\theta,\delta c_0} \; \frac12 \sum_{i=1}^N \bigl\| A_i \delta\theta + B_i \delta c_0 - r_i^k \bigr\|_{W_i}^2,
$$

где
$$
A_i = h_x J_\theta(t_i) + h_\theta,\qquad B_i = h_x J_{c_0}(t_i).
$$

В матричной форме, вводя $p = [\theta;\, c_0] \in \mathbb{R}^{n_\theta + n_x}$ и $\delta p = [\delta\theta;\,\delta c_0]$:
$$
\min_{\delta p} \; \frac12 \| J(p^k) \delta p - r(p^k) \|_W^2,
$$
где
$$
J(p) = \begin{bmatrix} A_1 & B_1 \\ \vdots & \vdots \\ A_N & B_N \end{bmatrix} \in \mathbb{R}^{(n_y N) \times (n_\theta + n_x)},\qquad
r(p) = \begin{bmatrix} r_1 \\ \vdots \\ r_N \end{bmatrix} \in \mathbb{R}^{n_y N},
$$
а $W = \operatorname{blkdiag}(W_1,\dots,W_N)$ — блочно-диагональная весовая матрица.

Решение $\delta p^*$ даёт обновление:
$$
p^{k+1} = p^k + \delta p^*.
$$
Это стандартный метод Гаусса–Ньютона.

## Multiple shooting: расширение на несколько интервалов

Разобьём временной интервал на $T$ непересекающихся промежутков (шутов):
$$
[t_0,t_f] = \bigcup_{j=1}^{T} [\tau_{j-1}, \tau_j], \quad \tau_0=t_0,\; \tau_T=t_f.
$$
В каждом шуте $j$ своё начальное состояние $c_j \in \mathbb{R}^{n_x}$. Условия непрерывности:
$$
x(\tau_j \mid \theta, c_j) = c_{j+1}, \quad j=1,\dots,T-1.
$$

Обобщённый вектор параметров:
$$
p = [\theta;\, c_1; c_2; \dots; c_T] \in \mathbb{R}^{n_\theta + n_x T}.
$$

Невязки измерений: для каждого измерения $y_i$, принадлежащего шуту $j$,
$$
r_i = y_i - h(x(t_i \mid \theta, c_j), \theta).
$$

Линеаризация даёт блочную матрицу $J$:

$$
J(p) =
\begin{bmatrix}
\Phi^1_{\theta}(t_1) & \Phi_{c_1}(t_1) & 0 & \cdots & 0 \\
\vdots & \vdots & 0 & \cdots & 0 \\
\Phi^1_{\theta}(t_{N_1}) & \Phi_{c_1}(t_{N_1}) & 0 & \cdots & 0 \\
\Phi^2_{\theta}(t_{N_1+1}) & 0 & \Phi_{c_2}(t_{N_1+1}) & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
\Phi^T_{\theta}(t_{N}) & 0 & 0 & \cdots & \Phi_{c_T}(t_{N})
\end{bmatrix}
$$
где $\Phi^j_{\theta}(t) = h_x J_\theta^j(t) + h_\theta$, $\Phi_{c_j}(t) = h_x J_{c_j}(t)$.

Размерность: $J(p) \in \mathbb{R}^{(n_y N) \times (n_\theta + n_x T)}$.

Условия непрерывности в линеаризованном виде:
$$
J_G(p) \, \delta p = h_{\text{cont}}(p),
$$
где
$$
J_G(p) =
\begin{bmatrix}
J_\theta^1(\tau_1) & J_{c_1}(\tau_1) & -I & 0 & \cdots & 0 \\
J_\theta^2(\tau_2) & 0 & J_{c_2}(\tau_2) & -I & \cdots & 0 \\
\vdots & \vdots & \vdots & \vdots & \ddots & \vdots \\
J_\theta^{T-1}(\tau_{T-1}) & 0 & 0 & 0 & \cdots & -I
\end{bmatrix},
$$
$$
h_{\text{cont}}(p) =
\begin{bmatrix}
c_2 - x(\tau_1 \mid \theta,c_1) \\
c_3 - x(\tau_2 \mid \theta,c_2) \\
\vdots \\
c_T - x(\tau_{T-1} \mid \theta,c_{T-1})
\end{bmatrix}.
$$
Размерность: $J_G(p) \in \mathbb{R}^{(n_x (T-1)) \times (n_\theta + n_x T)}$.

Таким образом, задача Multiple Shooting сводится к квадратичному программированию с линейными ограничениями-равенствами:
$$
\min_{\delta p} \; \frac12 \| J(p^k) \delta p - r(p^k) \|_W^2 \quad \text{s.t.} \quad J_G(p^k) \delta p = h_{\text{cont}}(p^k).
$$

## Решение задачи Multiple shooting методом множителей Лагранжа

Функция Лагранжа:
$$
L(\delta p, \lambda) = \frac12 \| J \delta p - r \|_W^2 + \lambda^T (J_G \delta p - h_{\text{cont}}).
$$

Условия оптимальности (ККТ):
$$
\begin{aligned}
\frac{\partial L}{\partial \delta p} &= J^T W (J \delta p - r) + J_G^T \lambda = 0,\\[4pt]
\frac{\partial L}{\partial \lambda} &= J_G \delta p - h_{\text{cont}} = 0.
\end{aligned}
$$

В матричной форме:
$$
\begin{bmatrix}
J^T W J & J_G^T \\
J_G & 0
\end{bmatrix}
\begin{bmatrix}
\delta p \\
\lambda
\end{bmatrix}
=
\begin{bmatrix}
J^T W r \\
h_{\text{cont}}
\end{bmatrix}.
$$

Эта система может быть решена, например, методом исключения или с помощью регуляризации (как в вашем коде). После нахождения $\delta p$ обновляем $p^{k+1} = p^k + \delta p$.

Таким образом, multiple shooting позволяет эффективно оценивать параметры и начальные условия на каждом интервале, обеспечивая непрерывность траектории через ограничения.

## Связь с кодом

- **`CompiledModel`** компилирует $f$, $h$ и их якобианы; **`VariationalIntegrator`** интегрирует $J_\theta$, $J_{c_0}$ (вариационные уравнения).
- **`get_jacobian_solution`** интегрирует расширенную систему и возвращает $x(t)$, $J_\theta(t)$, $J_{c_0}(t)$.
- **`MultipleShooting.shoot_rows`** собирает матрицы $J$ и $J_G$ и невязки $r$, $h_{\text{cont}}$ для одного батча.
- **Основной цикл** решает систему ККТ (с регуляризацией) и обновляет $\theta$ и $c_j$.

Это точная реализация описанного выше метода Гаусса–Ньютона с multiple shooting.

## Регуляризация

При решении задачи идентификации параметров методом Гаусса–Ньютона часто возникает проблема **плохой обусловленности** матрицы $J^T W J$ (особенно когда параметров много или данные содержат шум). Это приводит к неустойчивости оценок и переобучению.

### Байесовская интерпретация

Введём априорное распределение на параметры:
$$
\theta \sim \mathcal{N}(\bar\theta, \Lambda^{-1}), \qquad c_j \sim \mathcal{N}(\bar c_j, \Gamma_j^{-1}),
$$
где $\Lambda$, $\Gamma_j$ — матрицы точности (обратные ковариации). Тогда логарифм апостериорной плотности добавляет регуляризующие слагаемые:
$$
\frac12 \|\theta - \bar\theta\|_{\Lambda}^2 + \frac12 \sum_{j=1}^{T} \|c_j - \bar c_j\|_{\Gamma_j}^2.
$$

### Регуляризация в методе Гаусса–Ньютона

Целевая функция с регуляризацией:
$$
\min_{\delta p} \; \frac12 \| J \delta p - r \|_W^2 + \frac12 \| \delta p \|_{\text{diag}(\lambda)}^{2},
$$
где $\lambda$ — вектор коэффициентов регуляризации (обычно $\lambda_i \ge 0$). В матричной форме:
$$
\min_{\delta p} \; \frac12 \bigl( (J \delta p - r)^T W (J \delta p - r) + \delta p^T \operatorname{diag}(\lambda) \delta p \bigr).
$$

Это эквивалентно добавлению к матрице $J^T W J$ диагональной матрицы $\operatorname{diag}(\lambda)$:
$$
(J^T W J + \operatorname{diag}(\lambda)) \delta p = J^T W r.
$$


### Регуляризация в задаче с ограничениями (Multiple Shooting)

Для задачи с ограничениями-равенствами система ККТ преобразуется к виду:
$$
\begin{bmatrix}
J^T W J + \operatorname{diag}(\lambda) & J_G^T \\
J_G & 0
\end{bmatrix}
\begin{bmatrix}
\delta p \\ \lambda_{\text{lag}}
\end{bmatrix}
=
\begin{bmatrix}
J^T W r \\ h_{\text{cont}}
\end{bmatrix}.
$$


## Метод квадратичного штрафа (регуляризация блока ограничений)

При решении задачи multiple shooting методом Гаусса–Ньютона мы приходим к системе ККТ:

$$
\begin{bmatrix}
J^T W J & J_G^T \\
J_G & 0
\end{bmatrix}
\begin{bmatrix}
\delta p \\ \lambda
\end{bmatrix}
=
\begin{bmatrix}
J^T W r \\ h_{\text{cont}}
\end{bmatrix}.
$$

Блок $0$ в правом нижнем углу делает матрицу седловой и плохо обусловленной, что затрудняет численное решение. Для регуляризации применяется **метод квадратичного штрафа**: вместо задачи с ограничениями решается безусловная задача, в которой невыполнение ограничений штрафуется квадратично:

$$
\min_{\delta p} \; \frac12 \| J \delta p - r \|_W^2 + \frac{\rho}{2} \| J_G \delta p - h_{\text{cont}} \|^2, \qquad \rho > 0.
$$

Условия оптимальности:

$$
J^T W (J \delta p - r) + \rho\, J_G^T (J_G \delta p - h_{\text{cont}}) = 0.
$$

Введём вектор $\lambda = \rho\, (J_G \delta p - h_{\text{cont}})$ (играющий роль множителя Лагранжа). Тогда условия оптимальности эквивалентны системе

$$
\begin{aligned}
J^T W J\, \delta p + J_G^T \lambda &= J^T W r,\\
J_G \delta p - \tfrac{1}{\rho} \lambda &= h_{\text{cont}},
\end{aligned}
$$

или в матричной форме:

$$
\begin{bmatrix}
J^T W J & J_G^T \\
J_G & -\frac{1}{\rho} I
\end{bmatrix}
\begin{bmatrix}
\delta p \\ \lambda
\end{bmatrix}
=
\begin{bmatrix}
J^T W r \\ h_{\text{cont}}
\end{bmatrix}.
$$

Обратите внимание: штрафной член $\rho J_J_G^T J_G$ **не появляется** в верхнем левом блоке — он полностью учтён через переменную $\lambda$. При $\rho \to \infty$ система переходит в точную систему ККТ (нижний правый блок стремится к нулю).

### Связь с кодом

В реализации (`gauss_newton.adaptive.gn_step`) используется параметр $\mu = 1/\rho$:

$$
\begin{bmatrix}
J^T W J + \operatorname{diag}(\lambda_{\text{reg}}) & J_G^T \\
J_G & -\mu I
\end{bmatrix}
\begin{bmatrix}
\delta p \\ \lambda
\end{bmatrix}
=
\begin{bmatrix}
J^T W r \\ h_{\text{cont}}
\end{bmatrix},
$$

где $\operatorname{diag}(\lambda_{\text{reg}})$ — диагональная регуляризация из предыдущего раздела. Точная система ККТ восстанавливается при $\mu \to 0$; в коде $\mu$ стартует по кривизне и ужесточается по Пауэллу — только когда невязка стыковки не падает сама (`run_optimization_adaptive`, см. adaptive_regularization.ipynb).

Отличие от полного **метода расширенного Лагранжиана** (Augmented Lagrangian, Hestenes–Powell): там дополнительно явно накапливаются множители, $\lambda_{k+1} = \lambda_k + \rho\, (J_G \delta p - h_{\text{cont}})$, и штрафной вес можно не устремлять к бесконечности. В текущей реализации множители не накапливаются — это чистый квадратичный штраф с продолжением по $\mu$.

## Ковариационная матрица параметров

Ключевой момент: в multiple shooting невязки бывают двух видов, и это **разные объекты**.

- **Измерительные невязки** $R$ — случайны, это они несут шум $\varepsilon_i \sim \mathcal{N}(0,\Sigma_i)$ и определяют неопределённость оценки;
- **невязки непрерывности** $R_G$ — это **ограничения**: траектория обязана быть непрерывной, никакого шума в них нет. В решении они равны нулю (с точностью до $\mu$).

Поэтому объединять их в одну матрицу $J_{\text{full}} = [J; J_G]$ и писать
$\operatorname{Cov} = \hat\sigma^2 (J_{\text{full}}^T J_{\text{full}})^{-1}$ **нельзя**: это означало бы «непрерывность выполняется с точностью $\sigma$», то есть ограничение превращалось бы в наблюдение с произвольным весом.

### Правильная формула: ККТ-матрица

Оценка $\hat p$ решает задачу с ограничениями

$$
\min_p \ \|R(p)\|^2 \quad \text{s.t.} \quad G(p) = 0 .
$$

Возмущение данных $R \to R + \varepsilon$ сдвигает решение по касательному подпространству ограничений $\{\delta p : J_G\,\delta p = 0\}$. Линеаризуя условия ККТ, получаем ковариацию как **(1,1)-блок обратной ККТ-матрицы**:

$$
\boxed{\ \operatorname{Cov}(\hat p) \;=\; \hat\sigma^2 \left(
\begin{bmatrix} H & J_G^T \\ J_G & 0 \end{bmatrix}^{-1}
\right)_{11}, \qquad H = J^T W J . }
$$

При невырожденной $H$ этот блок равен

$$
H^{-1} - H^{-1} J_G^T \bigl( J_G H^{-1} J_G^T \bigr)^{-1} J_G H^{-1},
$$

то есть $H^{-1}$, **спроектированная** на касательное подпространство ограничений: ограничения не добавляют информации о шуме, они забирают степени свободы. Отсюда сразу два свойства, которых у прежней формулы не было:

- **инвариантность к масштабу ограничений**: замена $G \to \alpha G$ (то же самое ограничение) не меняет ни проектор, ни ковариацию;
- **согласованность с single shooting**: при $T=1$ ограничений нет и формула переходит в обычную $\hat\sigma^2 H^{-1}$; при $T>1$ она даёт **тот же** ответ, потому что это тот же самый оценщик.

### Оценка дисперсии шума

$$
\hat\sigma^2 = \frac{\|R\|^2}{\text{dof}}, \qquad
\text{dof} = n_{\text{meas}} + n_{\text{cont}} - n_p = n_{\text{meas}} - n_\theta - n_x .
$$

В числителе только **измерительная** невязка: строки стыковки не наблюдения и в остаточную сумму не входят. Знаменатель — число измерений минус число *свободных* неизвестных: из $n_p = n_\theta + n_x T$ неизвестных ограничения непрерывности связывают $n_x(T-1)$, свободны только $\theta$ и $c_1$.

### Маргинальная ковариация $\theta$

Интересны обычно физические параметры, а не начальные состояния шутов. Нужен $\theta$-блок, и он получается без явного обращения: правая часть $[\,I_{n_\theta};\,0\,]$, одна разреженная факторизация ККТ-матрицы (`splu`) и $n_\theta$ обратных подстановок. Это ровно та же матрица, которую решает `gn_step` при $\mu = 0$, поэтому отдельной машинерии не нужно (в коде — `NormalEquations.covariance_theta`, порядка 1.5 мс на задаче с 54 неизвестными).

Брать вместо этого $\hat\sigma^2 (H_{\theta\theta})^{-1}$ тоже нельзя: обращение только $\theta$-блока игнорирует корреляцию $\theta$ с начальными состояниями $c_j$.

### Что было не так раньше (и насколько)

До исправления в коде стояло $\operatorname{Cov} = \hat\sigma^2 (H + J_G^T J_G)^{-1}$ — непрерывность как наблюдение с весом 1. Проверки на Лотке–Вольтерре (`pytests/covariance_test.py`):

| проверка | прежняя формула | ККТ-формула |
|---|---|---|
| умножение $J_G$ на $10^3$ (то же ограничение) | СКО меняются в 2.4 раза | без изменений |
| сверка с single shooting | $+72\ldots+143\%$ | $-0.005\%$ |
| Монте-Карло, покрытие 95% интервалов | 100% (интервалы в 1.6–2.4 раза шире нужного) | 90–97% |

Завышение росло с числом шутов: $N_{\text{shoot}}=5$ — в 2.4 раза, $N_{\text{shoot}}=20$ — в 6.1 раза.

### Диагностика идентифицируемости

Из той же ковариации бесплатно получается корреляционная матрица
$\rho_{ij} = \operatorname{Cov}_{ij} / \sqrt{\operatorname{Cov}_{ii}\operatorname{Cov}_{jj}}$
(`NormalEquations.correlation_theta`). Значения $|\rho|$, близкие к 1, означают, что параметры различимы только в комбинации — «плоская долина» целевой функции. У Лотки–Вольтерры $\rho(\alpha,\beta) = 0.997$, $\rho(\gamma,\delta) = 0.90$, число обусловленности корреляционной матрицы $\approx 900$; на реальных данных то же самое наблюдается у пары $C_f, C_r$. Число обусловленности пишется в `hist['corr_cond']` по итерациям.

### Регуляризация

Для численной устойчивости к блоку $H$ добавляется малая диагональ $\varepsilon I$, $\varepsilon = 10^{-8}$ (на блок ограничений регуляризация не идёт — он не должен «размываться»). Если матрица всё же вырождена, используется псевдообратная.


## Доверительные интервалы для параметров

Для каждого параметра $p_i$ (компоненты вектора $\hat{p}$) строится **двусторонний доверительный интервал** с уровнем доверия $1-\alpha$ (например, $\alpha = 0.05$ для 95% интервала):

$$
\text{CI}_i = \hat{p}_i \;\pm\; t_{\alpha/2,\; \text{dof}} \cdot \text{se}_i,
$$

где:

- $\text{se}_i = \sqrt{\operatorname{Cov}(\hat{p})_{ii}}$ — стандартная ошибка $i$-го параметра,
- $t_{\alpha/2,\; \text{dof}}$ — критическое значение **t-распределения Стьюдента** с числом степеней свободы $\text{dof} = n_{\text{meas}} + n_{\text{cont}} - n_p$,
- $\alpha$ — уровень значимости (например, 0.05).

### Формула для границ

Нижняя граница:

$$
\text{CI}_{\text{low}, i} = \hat{p}_i - t_{\alpha/2,\; \text{dof}} \cdot \sqrt{\operatorname{Cov}(\hat{p})_{ii}}.
$$

Верхняя граница:

$$
\text{CI}_{\text{high}, i} = \hat{p}_i + t_{\alpha/2,\; \text{dof}} \cdot \sqrt{\operatorname{Cov}(\hat{p})_{ii}}.
$$

### Почему t-распределение?

При неизвестной истинной дисперсии шума и нормально распределённых остатках интервал, построенный с использованием t-распределения, даёт точный доверительный уровень для конечных выборок. При большом числе степеней свободы t-распределение приближается к нормальному.
